```python
import pandas as pd
import datetime
```

```python
df = pd.read_table('IVE.txt', sep=',', names=['Date', 'Time', 'Open', 'High', 'Low', 'Close', 'Volume'], header=None)

dt_col = df['Date'] + " " + df['Time']
f = lambda dt: datetime.datetime.strptime(dt, "%m/%d/%Y %H:%M")
df['Datetime'] = dt_col.apply(f)
df = df.drop(['Date', 'Time'], axis=1).set_index('Datetime')
df.tail()
```

```python
df.to_csv('IVE.csv')
```
